#### **B- Data Cleaning Part**

```python
# Convert Text in population size to numeric for visualization purposes
def extract_number(pop_size):

  # If the population size is in string format
  if isinstance(pop_size, str):

    if 'M' in pop_size:
      pop_size = float(pop_size.split('M')[0]) * 1000000  # 1M = 1,000,000
    elif 'k' in pop_size:
      pop_size = float(pop_size.split('k')[0]) * 1000     # 1k = 1,000
    elif 'B' in pop_size:
      pop_size = float(pop_size.split('B')[0]) * 1000000000      # 1B = 1,000, 000, 000
    else:
      pop_size = float(pop_size)

  # If the population size is in float or integer format
  elif isinstance(pop_size, (float, int)):
    pop_size = float(pop_size)

  # If unidentified format was detected then warn me
  else:
    print(f'Unpredefined type was detected for: {pop_size} , Detected type: {type(pop_size)}')

  return pop_size


# Create a copy and drop the 'country' column
population_df_copy = population_df.drop(columns=['country'], axis=1)


# Convert the values in each population size column to numeric values
for col in population_df_copy.select_dtypes(include='object').columns:
  population_df_copy[col] = population_df_copy[col].apply(extract_number)     # Extract the number part
  population_df_copy[col] = pd.to_numeric(population_df_copy[col], errors='coerce')            # Convert the column to numeric instead of object

# Return the country column back
population_df_copy['country'] = population_df['country']
```

### **2- Data Visualization & Exploration**

```python
# Get the years as a list to be used in plotting
years = list(population_df_copy.columns)
years.remove('country')

print('Years:', years)
```

```python
# See population size for each country, whether it decreases or increases over time
for index, row in population_df_copy.iterrows():

  data = row.drop('country').astype(float)

  # Plot the line plot
  sns.lineplot(x=years, y=data)
  plt.title(f'Population Growth Over Years For {row["country"]}')
  plt.xlabel('Year')
  plt.ylabel('Population Size')
  plt.show()


# Uzbekistan most likely would have a decline in the population size by 2060
# Country like USA (United States) is projected to have a linear increase in the population size
# Ukraine is having a decrease in population size (since 2000)
# Uganda & Tanzania started to have an increase in population size starting from 2000
# Sweden is projected to have a linear increase in population size since 1800
# Ireland had a very strange drop in population in the years [1930-1970] and started to grow back and is projected to reach a steady state around 2050
```

```python
# Plot the population count for 20 random counties across 20 random years
random_years_indicies = np.random.randint(low=0, high=len(years)-1, size=20)


# Compare the population Size for random countries in each random year
for index in random_years_indicies:

  year_name = years[index]

  # Randomly pick 50 countries
  n_countries = population_df_copy.shape[0]
  random_countries_indicies = np.random.permutation(range(0, n_countries))[:20]

  # Get filtered data
  data = population_df_copy.iloc[random_countries_indicies][['country', year_name]]
  x = data['country']
  y = data[year_name]

  # Plot the bar plot
  plt.figure(figsize=(20, 12))
  sns.barplot(x=x, y=y)
  plt.xlabel('Country')
  plt.ylabel('Population Size')
  plt.title(f'Population Count for Year {year_name}')
  plt.xticks(rotation = 45)
  plt.show()


# Some insights From the plots in the cell below
# Bangladesh had higher population count in 1993 than philipenes in Asia Content
# Turkey and Germany had competing population count in 2017
# In 2068 Ethiopia is projected to achieve higher population count than other african counties like Morroco and Kenya
```

```python
# Plot the population distribution for 20 random years
random_indicies = np.random.randint(low=0, high=len(years)-1, size=20)


for index in random_indicies:

  year_name = years[index]
  data = population_df_copy[year_name]

  # Plot the histogram
  sns.histplot(data)
  plt.title(f'Population Distribution for Year {year_name}')
  plt.show()


# Some Insights from the plots in the cell below
# You can see that most countries tend to have low populations
# Outliers are present (very few), these could be contributed to the presence of countries with large populations like India, China, Russia like seen previously
```

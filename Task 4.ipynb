```python
from selenium import webdriver
import pandas as pd
```

### **Part 1**

• Scrap books title, and prices from https://www.amazon.in/gp/bestsellers/books/ webpage and save it as a
CSV file.

#### **Initialize the web driver object**

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.amazon.in/gp/bestsellers/books/")
```

#### **Extract the Books' Data**

```python
# Get Books' data
books_title = driver.find_elements("xpath", "//span/div[@class=\"_cDEzb_p13n-sc-css-line-clamp-1_1Fn1y\"]")       
books_author = driver.find_elements("xpath", "//a[@class=\"a-size-small a-link-child\"]") 
books_rating = driver.find_elements("xpath", "//span[@class=\"a-icon-alt\"]")
books_items_bought = driver.find_elements("xpath", "//a[@class=\"a-link-normal\"]/span[@class=\"a-size-small\"]")
books_price = driver.find_elements("xpath", "//span[@class=\"_cDEzb_p13n-sc-price_3mJ9Z\"]")                
```

#### **Map the extracted data to python dictionary**

```python
# Create a dictionary to store Book Data
book_data = {'Title': [],
             'Author': [],
             'Rating': [],
             'Quantity Bought': [],
             'Price': []}
```

```python
for idx, zipped_book_info in enumerate(zip(books_title, books_author, books_rating, books_items_bought, books_price)):

    book_title, book_author, book_rating, items_bought, book_price = zipped_book_info     # Get the book's data

    # Visualize the extracted Data
    print(f'##################################### Book {idx+1} Before Refinning Data #####################################')
    print(f'{idx+1}. Title:', book_title.text)
    print(f'{idx+1}. Author:', book_author.text)
    print(f'{idx+1}. Rating:', book_rating.text)
    print(f'{idx+1}. Items Bought:', items_bought.text)
    print(f'{idx+1}. Price:', book_price.text)

    # Refine the data to get the exact information required
    book_title = book_title.text
    book_author = book_author.text
    items_bought = items_bought.text
    book_rating = book_rating.get_attribute("innerText").strip()
    book_price = book_price.text


    print(f'##################################### Book {idx+1} After Refinning Data #####################################')
    print(f'{idx+1}. Title:', book_title)
    print(f'{idx+1}. Author:', book_author)
    print(f'{idx+1}. Items Bought:', items_bought)
    print(f'{idx+1}. Rating:', book_rating)
    print(f'{idx+1}. Price:', book_price)


    # Save the Data in the dictionary
    book_data['Title'].append(book_title)
    book_data['Author'].append(book_author)
    book_data['Rating'].append(book_rating)
    book_data['Quantity Bought'].append(items_bought)
    book_data['Price'].append(book_price)
```

#### **Convert the python dictionary to DataFrame**

```python
book_data_df = pd.DataFrame(book_data)
```

```python
book_data_df.head()
```

#### **Save the DataFrame AS CSV File**

```python
book_data_df.to_csv('book_data.csv', index=False)
```

### **Part 2**

• Your task is to scrape the top 200 movies with their lifetime collection and  release data and save it in a CSV file. The website we will be using is box office mojo  https://www.boxofficemojo.com/chart/top_lifetime_gross/?area=XWW

#### **Initialize the web driver object**

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.boxofficemojo.com/chart/top_lifetime_gross/?area=XWW")
```

#### **Extract the Movies' Data**

```python
# Get Movies' data
movies_rank = driver.find_elements("xpath", "//tr/td[@class=\"a-text-right mojo-header-column mojo-truncate mojo-field-type-rank\"]") 
movies_title = driver.find_elements("xpath", "//tr/td[@class=\"a-text-left mojo-field-type-title\"]")       
movies_lifetime_gross = driver.find_elements("xpath", "//tr/td[@class=\"a-text-right mojo-field-type-money\"]")
movies_year = driver.find_elements("xpath", "//tr/td[@class=\"a-text-left mojo-field-type-year\"]")
```

#### **Map the extracted data to python dictionary**

```python
# Create a dictionary to store Movie Data
movie_data = {'Title': [],
              'Rank': [],
              'Lifetime Gross': [],
              'Year': []}
```

```python
for idx, zipped_movie_info in enumerate(zip(movies_rank, movies_title, movies_lifetime_gross, movies_year)):

    movie_rank, movie_title, movie_lifetime_gross, movie_year = zipped_movie_info     # Get the movie's data

    # Visualize the extracted Data
    print(f'##################################### Movie {idx+1} Data #####################################')
    print(f'{idx+1}. Title:', movie_title.text)
    print(f'{idx+1}. Rank:', movie_rank.text)
    print(f'{idx+1}. Lifetime Gross:', movie_lifetime_gross.text)
    print(f'{idx+1}. Year:', movie_year.text)

    # Refine the data to get the exact information required
    movie_rank = movie_rank.text
    movie_title = movie_title.text
    movie_lifetime_gross = movie_lifetime_gross.text
    movie_year = movie_year.text


    # Save the Data in the dictionary
    movie_data['Title'].append(movie_title)
    movie_data['Rank'].append(movie_rank)
    movie_data['Lifetime Gross'].append(movie_lifetime_gross)
    movie_data['Year'].append(movie_year)
```

#### **Convert the python dictionary to DataFrame**

```python
movie_data_df = pd.DataFrame(movie_data)
```

```python
movie_data_df.head()
```

#### **Save the DataFrame AS CSV File**

```python
movie_data_df.to_csv('movie_data.csv', index=False)
```
